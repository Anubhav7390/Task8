```python
# droping null values from postal columns

df.dropna(subset = ['Postal Code'], inplace = True)
```

```python
df.isnull().sum()
```

```python
# It aggregates the data by default.
# errorbar or ci(confidence interval) to remove little line in top of bars.
# ci parameter has been deprecated thats why use errorbar works same.


ax = sns.barplot(x = "Category", y = "Sales", data = df, color = "pink", errorbar = None)
plt.title(" Sales by Category")
plt.xticks(rotation = 45)
ax.bar_label(ax.containers[0], fmt='%.2f',label_type='center', color='black') # fmt = %.2f for 2 decimal places
plt.show()
```

```python
# It does not aggregate.


plt.bar(df['Category'], df['Sales'])
plt.title("Category by Sales")
plt.show()
```

```python
agg_data = df.groupby('Category')['Sales'].mean().round(2)
agg_data
```

```python
# to get same result in plt.

agg_data = df.groupby('Category', as_index=False)['Sales'].mean()
plt.bar(agg_data['Category'], agg_data['Sales'])
plt.show()
```

```python
# sales by Region

o = ['East', 'West', 'Central', 'South' ]
ax = sns.barplot(x = "Region", y = "Sales", data = df, order = o, color = "pink", errorbar = None)
ax.bar_label(ax.containers[0], fmt = "%.2f", label_type = "center", color = 'black')
plt.title("Sales by Region")
plt.xticks(rotation = 45)
plt.show()
```

```python
ax = sns.barplot(x = "Ship Mode", y = "Sales", data = df, color = "pink", errorbar = None)
ax.bar_label(ax.containers[0], fmt = "%.2f", label_type = "center", color = 'black')
plt.title("Sales by Ship Mode")
plt.xticks(rotation = 45)
plt.show()
```
